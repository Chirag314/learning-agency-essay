# 01 — EDA

Real exploratory analysis of the 17,307-essay AES2 train set. See `reports/eda_report.md` for the full writeup with embedded charts; this notebook is the runnable source for that report.

```
import sys
sys.path.insert(0, '..')
import pandas as pd

df = pd.read_csv('../data/train.csv')
print(df.shape)
df.head()
```

```
df['score'].value_counts().sort_index()
```

```
df['word_count'] = df['full_text'].str.split().str.len()
df['char_len'] = df['full_text'].str.len()
df[['word_count', 'char_len']].describe()
```

```
print('word count vs score correlation:', df['word_count'].corr(df['score']))
```

Full chart generation: `eda/eda_essays.py`. Key finding: word count correlates 0.69 with score — the strongest simple signal in the dataset (see `reports/eda_report.md`).
